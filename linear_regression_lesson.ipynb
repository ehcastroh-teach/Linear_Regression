#### PART 0

## **SUMMARY STATISTICS** from Scratch


Before reaching for a library, we compute the building blocks by hand on a small toy dataset. This forces
familiarity with what `np.mean`, `np.cov`, and `np.var` actually calculate - which matters when interpreting
model output later.

The key quantity we are building toward is **covariance** - it captures how much two variables move together.
If $\text{Cov}(X, Y)$ is positive, higher $X$ tends to mean higher $Y$. If negative, they move in opposite
directions. Linear regression is essentially a normalized version of this idea:

$$\text{slope} = \frac{\text{Cov}(X, Y)}{\text{Var}(X)}$$

where slope $= \frac{\text{Cov}(X, Y)}{\text{Var}(X)}$ tells us: for every one-unit increase in $X$, how much does $Y$ change on average?
Dividing by $\text{Var}(X)$ normalizes the covariance so the slope is in units of "$Y$ per unit of $X$",
not in units of "$X \times Y$".

___

**Note:** This dataset is intentionally small and artificial. The noise is high - $X$ and $Y$ are not strongly
correlated here - because the goal is to practice the formulas, not to build an accurate model.

___


Consider the following 9-point dataset:

| X  | Y  |
|----|----|
| 7  | 12 |
| 4  | 23 |
| 6  | 64 |
| 3  | 96 |
| 5  | 23 |
| 8  | 52 |
| 1  | 67 |
| 13 | 52 |
| 10 | 68 |


````python
x = np.array([7, 4, 6, 3, 5, 8, 1, 13, 10])
y = np.array([12, 23, 64, 96, 23, 52, 67, 52, 68])
````

#### 0.1 Expected Value E[X]


The **expected value** (or mean) of $X$ is the arithmetic average:

$$E[X] = \frac{1}{n} \sum_{i=1}^{n} x_i$$

It is the center of mass of the distribution - if you picked a random point from this dataset repeatedly,
$E[X]$ is the long-run average of what you would get.


````python
q1a = np.mean(x)
print(f"E[X] = {q1a}")
````

#### 0.2 Expected Value E[Y]


Same calculation for $Y$. We will need $E[Y]$ to compute the regression intercept:

$$E[Y] = \frac{1}{n} \sum_{i=1}^{n} y_i$$


````python
q1b = np.mean(y)
print(f"E[Y] = {q1b}")
````

#### 0.3 Covariance Cov(X, Y)


Covariance measures how much $X$ and $Y$ move together. The formula is:

$$\text{Cov}(X, Y) = \frac{1}{n-1} \sum_{i=1}^{n} (x_i - E[X])(y_i - E[Y])$$

Each term $(x_i - E[X])(y_i - E[Y])$ is positive when both $x_i$ and $y_i$ are on the same side of their
respective means, and negative when they are on opposite sides. Summing these tells us the dominant direction
of co-movement. We divide by $n-1$ (not $n$) to get an unbiased estimate of the population covariance from a sample.

`np.cov` returns a 2x2 matrix; the off-diagonal entry $[0][1]$ is $\text{Cov}(X, Y)$.


````python
q1c = np.cov(x, y)[0][1]
print(f"Cov(X, Y) = {q1c:.4f}")
````

#### 0.4 Expected Value E[X^2]


$E[X^2]$ is the mean of the squared values - not the square of the mean. The two differ unless variance is zero.
This quantity appears in some derivations of the variance formula and in higher-moment calculations.

$$E[X^2] = \frac{1}{n} \sum_{i=1}^{n} x_i^2$$


````python
q1d = np.mean(x**2)
print(f"E[X^2] = {q1d}")
````

#### PART 1

## **SINGLE-VARIABLE** Predictor


You are now a real estate broker in Caltopia, a town with many MRT stations and convenience stores. You want
to price houses fairly. Your dataset covers 414 past real estate transactions with features like house age,
distance to transit, and number of nearby stores.

In this part, we predict house price from **house age alone** - one input, one output. The closed-form
solution for simple linear regression is:

$$\text{slope} = \frac{\text{Cov}(X_1, Y)}{\text{Var}(X_1)}, \quad \text{intercept} = E[Y] - \text{slope} \cdot E[X_1]$$

where slope $= \frac{\text{Cov}(X_1, Y)}{\text{Var}(X_1)}$ tells us: for every one additional year of house age, how much does
the price per unit area change on average? And intercept $= E[Y] - \text{slope} \cdot E[X_1]$ is the predicted
price for a hypothetical house of age zero - not a realistic house, but a necessary mathematical anchor.

This is the **analytic** solution to linear regression - no gradient descent, no optimization loop. We derive
the slope directly from two statistics computed from the data.


#### 1.1 Load the Dataset

We load the dataset and extract each feature column as a NumPy array. The target variable is `Y house price of unit area`.


````python
dataset = pd.read_csv("real_estate_data.csv")
dataset.head()
````

````python
x1 = dataset["X1 house age"].to_numpy()
x2 = dataset["X2 distance to the nearest MRT station"].to_numpy()
x3 = dataset["X3 number of convenience stores"].to_numpy()
x4 = dataset["X4 latitude"].to_numpy()
x5 = dataset["X5 longitude"].to_numpy()
y  = dataset["Y house price of unit area"].to_numpy()

print(f"Dataset shape: {dataset.shape}  ({len(y)} houses, {dataset.shape[1]} columns)")
````

#### 1.2 Summary Statistics for the Predictor

We need four statistics to build the single-variable predictor: $E[X_1]$, $E[Y]$, $\text{Var}(X_1)$, and
$\text{Cov}(X_1, Y)$. Each formula below uses sample statistics (dividing by $n-1$) to get unbiased estimates
from our 414-point sample.


````python
# Expected age of a random house from this dataset
q2a = np.mean(x1)

# Expected price per unit area
q2b = np.mean(y)

# Sample variance in house age (ddof=1 for n-1 denominator)
q2c = np.var(x1, ddof=1)

# Sample covariance between age and price
q2d = np.cov(x1, y)[0][1]

print(f"E[X1 house age]      = {q2a:.4f} years")
print(f"E[Y price/unit area] = {q2b:.4f}")
print(f"Var(X1)              = {q2c:.4f}")
print(f"Cov(X1, Y)           = {q2d:.4f}")
````

#### 1.3 Build the Linear Predictor

With slope and intercept computed from the sample statistics, we define the predictor function.
Note that every error in $\text{Cov}$ or $\text{Var}$ carries forward into the slope - a wrong variance
estimate would shift every prediction. This is intentional: it shows that a regression model is only as
reliable as the summary statistics feeding it.


````python
def linear_predictor(X):
    """Predict house price per unit area from house age X.
    
    slope     = Cov(X1, Y) / Var(X1)
    intercept = E[Y] - slope * E[X1]
    """
    slope = q2d / q2c
    intercept = q2b - slope * q2a
    return intercept + slope * X

slope_val = q2d / q2c
intercept_val = q2b - slope_val * q2a
print(f"slope     = {slope_val:.4f}  (price change per additional year of house age)")
print(f"intercept = {intercept_val:.4f} (predicted price for a house of age 0)")

# Predictions for a sample of house ages
test_ages = np.array([6.0, 32.0, 9.0, 4.0, 1.0])
predictions = linear_predictor(test_ages)
for age, price in zip(test_ages, predictions):
    print(f"  Age {age:5.1f} years  ->  predicted price/unit area: {price:.2f}")
````

#### PART 2

## **MULTI-VARIABLE** Predictor via the Normal Equation


One feature is rarely enough. Part 1 showed that house age alone is a weak predictor. Here we add
$x_2$ (distance to MRT station) and $x_3$ (number of nearby convenience stores) as additional inputs.

With multiple inputs, the scalar formula no longer applies - we need matrix algebra. The **normal equation**
gives the exact, analytic solution for the weight vector $W$ that minimizes the sum of squared prediction errors:

$$W = (X^\top X)^{-1} X^\top Y$$

where:
- $X$ is the $n \times p$ design matrix ($n$ = 414 houses, $p$ = 3 features)
- $Y$ is the $n \times 1$ vector of house prices
- $X^\top X$ is the $p \times p$ matrix of feature cross-products (invertible when features are linearly independent)
- $W$ is the $p \times 1$ weight vector - each entry scales one feature

The term $(X^\top X)^{-1} X^\top$ is the Moore-Penrose pseudoinverse of $X$. When $n \gg p$ and columns
are not linearly dependent, this inverse exists and the result is the unique least-squares solution.

___

**Note:** The normal equation is exact and has no hyperparameters - it is not iterative and does not require
a learning rate. Its limitation is computational: inverting $X^\top X$ costs $O(p^3)$, which becomes
expensive when $p$ is large (thousands of features). For large $p$, gradient descent is preferred.

___


#### 2.1 Construct the Design Matrix

We stack the three feature columns into a matrix $X$ where each **row** is one house and each **column**
is one feature. `Y` is reshaped to a column vector so matrix multiplication is well-defined.


````python
# Stack features as columns: shape (414, 3)
X = np.column_stack((x1, x2, x3))

# Reshape target to column vector: shape (414, 1)
Y = y.reshape(-1, 1)

print(f"X shape: {X.shape}  (rows=houses, cols=features)")
print(f"Y shape: {Y.shape}")
print(f"\nFirst 3 rows of X:")
print(X[:3])
````

#### 2.2 Solve for the Weight Matrix

We apply the normal equation step by step:

1. $X^\top X$ - a $3 \times 3$ matrix of feature inner products
2. $(X^\top X)^{-1}$ - its matrix inverse (requires linearly independent columns)
3. $(X^\top X)^{-1} X^\top$ - multiply by the transposed feature matrix
4. $(X^\top X)^{-1} X^\top Y$ - multiply by the target vector to get weights


````python
W = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(Y)

print("Weight matrix W:")
print(f"  w1 (house age weight):     {W[0][0]:.6f}")
print(f"  w2 (MRT distance weight):  {W[1][0]:.6f}")
print(f"  w3 (convenience stores):   {W[2][0]:.6f}")
print()
print("Interpretation:")
print(f"  Each additional year of house age changes price by {W[0][0]:.4f} units/area")
print(f"  Each additional meter from MRT changes price by   {W[1][0]:.6f} units/area")
print(f"  Each additional convenience store changes price by {W[2][0]:.4f} units/area")
````

#### 2.3 Build the Multi-Variable Predictor

The predictor unpacks the three weights from $W$ and computes a price for any new combination of inputs.
Note the weight ordering must match the column ordering in $X$: $w_1$ for age, $w_2$ for MRT distance,
$w_3$ for convenience stores.


````python
def predictor(house_age, distance_to_mrt, num_convenience_stores):
    """Predict house price per unit area from three features.
    
    Weight order matches X column order: [age, MRT dist, stores]
    """
    w1 = float(W[0])
    w2 = float(W[1])
    w3 = float(W[2])
    return w1 * house_age + w2 * distance_to_mrt + w3 * num_convenience_stores

# Example prediction
house_age = 32
distance_to_mrt = 123
num_stores = 5

predicted_price = predictor(house_age, distance_to_mrt, num_stores)
print(
    f"House: {house_age} years old, {distance_to_mrt}m from MRT, {num_stores} nearby stores\n"
    f"  Predicted price per unit area: {predicted_price:.2f}"
)
````
